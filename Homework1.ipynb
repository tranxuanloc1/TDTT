**Bài 2:** Hình chữ nhật a, b. Tính diện tích còn lại khi đục lỗ hình tròn bán kính a/2. (pi=3.14, 2 chữ số thập phân)

```python
a=int(input())
b=int(input())
pi=3.14
S_trung_tam=pi*((a/2)**2)
S_hcn=a*b
S=round(S_hcn-S_trung_tam,2)
print(S)
```

**Bài 6:** Kiểm tra 3 cạnh tam giác, tính diện tích

```python
a=int(input())
b=int(input())
c=int(input())
if a+b<c or a+c<b or b+c<a:
    print("Khong phai 3 canh cua tam giac")
else:
    p=(a+b+c)/2
    S=(p*(p-a)*(p-b)*(p-c))**0.5
    print(S)
```
